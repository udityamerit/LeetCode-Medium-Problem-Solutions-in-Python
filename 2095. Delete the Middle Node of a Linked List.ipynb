```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def deleteMiddle(self, head: ListNode | None) -> ListNode | None:
        l = 0
        temp = head
        while temp:
            temp = temp.next
            l+=1

        l= l//2

        if head is None or head.next is None:
            return None

        tmep = head
        idx=0
        while tmep and idx<l-1:
            tmep = tmep.next
            idx+=1

        if tmep and tmep.next:
            tmep.next = tmep.next.next

        return head
```
