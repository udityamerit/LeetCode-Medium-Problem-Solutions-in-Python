#### Solution 1 in python

```python
class Solution:
    def minimumPushes(self, word: str) -> int:
        d = {}
        for i in word:
            if i in d:
                d[i]+=1
            else:
                d[i]=1
        # print(d)
        l =[]
        for key, values in d.items():
            l.append(values)
            # print(d.values)
        l.sort(reverse=True)
        push = 0
        for i in range(len(l)):
            push+= ((i//8)+1)*l[i]

        return push
```
