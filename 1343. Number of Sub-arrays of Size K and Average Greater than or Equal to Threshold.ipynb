```python
class Solution:
    def numOfSubarrays(self, arr: List[int], k: int, threshold: int) -> int:
        count = 0
        cur_sum = sum(arr[:k])
        if cur_sum//k>=threshold:
            count+=1
        tar_sum = threshold*k
        for i in range(k, len(arr)):
            cur_sum = cur_sum+arr[i]-arr[i-k]
            if cur_sum//k>=threshold:
                count+=1
        
        return count
```
