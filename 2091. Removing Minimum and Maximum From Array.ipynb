```python
from typing import List

class Solution:
    def minimumDeletions(self, nums: List[int]) -> int:
        n = len(nums)
        if n <= 2:
            return n
            
        # Find indices of the minimum and maximum elements
        min_val = min(nums)
        max_val = max(nums)
        
        mi = nums.index(min_val)
        mx = nums.index(max_val)
        
        i = min(mi, mx)
        j = max(mi, mx)
        
        from_front = j + 1
        
        from_back = n - i
        
        from_both = (i + 1) + (n - j)
        
        return min(from_front, from_back, from_both)
```
