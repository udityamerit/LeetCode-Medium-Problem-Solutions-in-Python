#### Solution 1 in python

```python
class Solution:
    def valueAfterKSeconds(self, n: int, k: int) -> int:
        l = [1 for i in range(n)]
        print(l)
        while k:
            ans =0
            a=[]
            for i in l: 
                ans+=i
                a.append(ans)
            l = a
            k-=1
        return l[-1]%(10**9+7)
```
