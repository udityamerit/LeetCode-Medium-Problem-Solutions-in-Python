#### Solution 1 in python

```python
class Solution:
    def passwordStrength(self, password: str) -> int:
        s = "!@#$"
        st  =0

        seen = list(set(password))
        # print(seen)
        for i in seen:
            if i.islower():
                st +=1
            elif i.isupper():
                st+=2
            elif i.isnumeric():
                st+=3
            elif i in s:
                st+=5
        return st
```
