```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def sortList(self, head: ListNode | None) -> ListNode | None:
        arr =[]
        temp = head
        while temp:
            arr.append(temp.val)
            temp = temp.next

        arr.sort()
        head =None
        tail = None
        for i in arr:
            node = ListNode(i)
            if head is None:
                head = node
                tail = node
            else:
                tail.next =node
                tail =node

        return head
```
