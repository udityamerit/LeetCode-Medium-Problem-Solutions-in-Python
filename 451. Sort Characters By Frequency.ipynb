#### Solution 1 in python

```python
class Solution:
    def frequencySort(self, s: str) -> str:
        fre = {}
        for i in s:
            if i in fre:
                fre[i] +=1
            else:
                fre[i]=1
        
        s = dict(sorted(fre.items(), key=lambda item: item[1], reverse=True))
        print(s)
        ans = ""
        for key, values in s.items():
            ans = ans + key*values

        # print(ans)  
        return ans    
            
```
