#### Solution 1 in python

```python
class Solution:
    def reorderList(self, head: ListNode | None) -> None:
        """
        Do not return anything, modify head in-place instead.
        """
        if not head or not head.next:
            return

        arr1 = []
        temp = head
        while temp:
            arr1.append(temp)  
            temp = temp.next
            
        
        arr2 = []
        l = 0
        h = len(arr1) - 1
        while l < h:
            arr2.append(arr1[l])
            l += 1
            arr2.append(arr1[h])
            h -= 1
            
        if len(arr1) % 2 != 0:
            arr2.append(arr1[len(arr1) // 2])
            
       
        for i in range(len(arr2) - 1):
            arr2[i].next = arr2[i + 1]
            
       
        arr2[-1].next = None
```
