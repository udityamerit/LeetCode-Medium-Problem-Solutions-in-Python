#### Solution 1 in python in using list and linkedlist

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def mergeNodes(self, head: ListNode | None) -> ListNode | None:
        l = []
        temp = head
        while temp:
            l.append(temp.val)
            temp = temp.next
        
        c = 0
        m = []
        for i in range(len(l)):
            if l[i] ==0:
                m.append(c)
                c = 0
            else:
                c+=l[i]

        m.pop(0)
        head = None
        tail = None
        for i in m:
            node = ListNode(i)
            if head is None:
                head = node
                tail = node
            else:
                tail.next = node
                tail = node
        
        return head
```
