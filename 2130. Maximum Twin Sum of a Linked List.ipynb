```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def pairSum(self, head: ListNode | None) -> int:
        ans = []
        sc = 0
        temp = head
        while temp:
            ans.append(temp.val)
            temp =temp.next

        left = 0
        right = len(ans)-1
        while left<right:
            sc = max(sc, ans[left]+ans[right])
            left+=1
            right-=1

        return sc
```
