#### Solution 1 in python

```python
class Solution:
    def findAnagrams(self, s: str, p: str) -> list[int]:
        ans = []
        k = len(p)
        sorted_p = ''.join(sorted(p))
        
        for i in range(len(s) - k + 1):
            
            if ''.join(sorted(s[i:i+k])) == sorted_p:
                ans.append(i)
        return ans   
```
