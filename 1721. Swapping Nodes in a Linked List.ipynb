#### Solution 1 in python

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def swapNodes(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        arr = []
        temp = head
        while temp:
            arr.append(temp.val)
            temp = temp.next

        arr[k-1], arr[-k]= arr[-k], arr[k-1]
        head=None
        tail = None
        for i in range(len(arr)):
            node = ListNode(arr[i])
            if head is None:
                head =node
                tail = node
            else:
                tail.next = node
                tail=node
        
        return head

```
