#### Solution 1 in python

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        l = 0
        r = len(height)-1
        maxwater = 0
        while(l<r):
            curwater = min(height[l], height[r])*(r-l)
            maxwater = max(maxwater, curwater)

            if height[l]<height[r]:
                l+=1
            else:
                r-=1
        return maxwater
```
