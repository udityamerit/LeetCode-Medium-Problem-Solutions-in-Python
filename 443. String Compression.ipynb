#### Solution 1 in python

```python
class Solution:
    def compress(self, chars: list[str]) -> int:
        if not chars:
            return 0
            
        l = []
        c = 1
        
        for i in range(len(chars) - 1):
            if chars[i] != chars[i + 1]:
                l.append(chars[i])
                if c > 1:                  
                    for digit in str(c):   
                        l.append(digit)
                c = 1
            else:
                c += 1
                
        l.append(chars[-1])
        if c > 1:
            for digit in str(c):
                l.append(digit)
        for i in range(len(l)):
            chars[i] = l[i]
            
        return len(l)
```
