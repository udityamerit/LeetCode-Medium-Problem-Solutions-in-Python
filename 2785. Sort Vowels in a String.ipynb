#### Solution 1 in python

```python
class Solution:
    def sortVowels(self, s: str) -> str:
        s_list = list(s)
        v = "aeiouAEIOU"
        ex = sorted([char for char in s_list if char in v])
        vowel_index = 0
        for i in range(len(s_list)):
            if s_list[i] in v:
                s_list[i] = ex[vowel_index]
                vowel_index += 1
                
        return "".join(s_list)
```
