#### Solution 1 in python

```python
class Solution:
    def processStr(self, s: str) -> str:
        op = ""
        for i in s:
            if i == "#":
                op = op * 2
            elif i == "%":
                
                op = op[::-1] 
            elif i == "*":
                if op: 
                    op = op[:-1]
            else:
                op = op + i

        return op
```
