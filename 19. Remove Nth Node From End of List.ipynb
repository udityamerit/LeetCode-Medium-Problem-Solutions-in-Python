#### Solution 1 in python

```python
class Solution:
    def removeNthFromEnd(self, head: ListNode | None, n: int) -> ListNode | None:
        if head is None:
            return head
            
        l = 0
        temp = head
        while temp:
            temp = temp.next
            l += 1

        posi = l - n
        
        if posi == 0:
            return head.next

        ind = 0
        temp = head
        while temp is not None and ind < posi - 1:
            temp = temp.next
            ind += 1
            
        if temp and temp.next:
            temp.next = temp.next.next

        return head
```
